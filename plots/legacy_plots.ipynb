```python
import matplotlib.pyplot as plt
import numpy as np

classes = ["PICKUP", "SUV", "BTR70", "BRDM2", "BMP2", "T72", "ZSU23", "2S3", "MTLB", "D20"]

fusion_ssd_seen = [0.4396, 0.5505, 0.6514, 0.6034, 0.7331, 0.7002, 0.6839, 0.7047, 0.5851, 0.1299]
fusion_ssd_unseen = [0.0064, 0.0732, 0.1472, 0.0695, 0.1688, 0.1278, 0.0665, 0.2157, 0.0451, 0.0219]

vlc_seen = [0.4998, 0.5535, 0.6290, 0.5751, 0.6831, 0.7149, 0.6563, 0.6672, 0.5449, 0.4946]
vlc_unseen = [0.0103, 0.1012, 0.1832, 0.0812, 0.2361, 0.2267, 0.1156, 0.2584, 0.0659, 0.0176]

PINK = "#E91E63"
BLUE = "#1976D2"

def make_radar(path, vals_a, label_a, vals_b, label_b, rmax):
    vals_a = [v * 100 for v in vals_a]
    vals_b = [v * 100 for v in vals_b]
    N = len(classes)
    angles = np.linspace(0, 2 * np.pi, N, endpoint=False).tolist()
    angles += angles[:1]
    vals_a_plot = vals_a + vals_a[:1]
    vals_b_plot = vals_b + vals_b[:1]

    fig = plt.figure(figsize=(9, 9))
    ax = plt.subplot(111, polar=True)

    ax.plot(angles, vals_a_plot, linewidth=2.5, label=label_a,
            color=PINK, marker="o", markersize=8)
    ax.fill(angles, vals_a_plot, alpha=0.15, color=PINK)

    ax.plot(angles, vals_b_plot, linewidth=2.5, label=label_b,
            color=BLUE, marker="s", markersize=8)
    ax.fill(angles, vals_b_plot, alpha=0.15, color=BLUE)

    ax.set_xticks(angles[:-1])
    ax.set_xticklabels(classes, fontsize=24, fontweight="bold")
    ax.set_ylim(0, rmax)
    yticks = list(range(10, rmax + 1, 10))
    ax.set_yticks(yticks)
    ax.set_yticklabels([str(v) for v in yticks], fontsize=13)
    ax.tick_params(axis="both", labelsize=16)
    ax.legend(loc="upper right", bbox_to_anchor=(1.25, 1.12),
              fontsize=20, framealpha=0.9)

    plt.tight_layout()
    plt.savefig(path, dpi=300, bbox_inches="tight")
    plt.close(fig)

seen_path = "atr_seen_vlc_vs_fusionssd_radar.pdf"
unseen_path = "atr_unseen_vlc_vs_fusionssd_radar.pdf"

make_radar(
    seen_path,
    vlc_seen,
    "VLC Fusion",
    fusion_ssd_seen,
    "Fusion SSD",
    rmax=80,
)
make_radar(
    unseen_path,
    vlc_unseen,
    "VLC Fusion",
    fusion_ssd_unseen,
    "Fusion SSD",
    rmax=30,
)

print(seen_path)
print(unseen_path)
```

```python
import matplotlib.pyplot as plt
import numpy as np

labels = ["Vehicle-L1", "Vehicle-L2", "Cyclist-L1", "Cyclist-L2", "Ped.-L1", "Ped.-L2"]

seen_fusionssd = [19.7, 16.9, 21.5, 19.7, 37.2, 32.4]
seen_vlc10 = [31.04, 26.80, 24.55, 22.54, 43.81, 38.64]

unseen_fusionssd = [21.5, 18.8, 31.1, 29.4, 40.4, 37.4]
unseen_vlc10 = [32.01, 28.3, 38.4, 36.4, 46.2, 43.0]

PINK = "#E91E63"
BLUE = "#1976D2"

def radar_plot(path, vals_a, label_a, vals_b, label_b, rmax):
    N = len(labels)
    angles = np.linspace(0, 2*np.pi, N, endpoint=False).tolist()
    angles += angles[:1]
    vals_a_plot = vals_a + vals_a[:1]
    vals_b_plot = vals_b + vals_b[:1]

    fig = plt.figure(figsize=(9, 9))
    ax = plt.subplot(111, polar=True)

    ax.plot(angles, vals_a_plot, linewidth=2.5, label=label_a,
            color=PINK, marker="o", markersize=8)
    ax.fill(angles, vals_a_plot, alpha=0.15, color=PINK)

    ax.plot(angles, vals_b_plot, linewidth=2.5, label=label_b,
            color=BLUE, marker="s", markersize=8)
    ax.fill(angles, vals_b_plot, alpha=0.15, color=BLUE)

    ax.set_xticks(angles[:-1])
    ax.set_xticklabels(labels, fontsize=24, fontweight="bold")
    ax.set_ylim(0, rmax)
    yticks = list(range(5, rmax + 1, 5))
    ax.set_yticks(yticks)
    ax.set_yticklabels([str(v) for v in yticks], fontsize=13)
    ax.tick_params(axis="both", labelsize=16)
    ax.legend(loc="upper center", bbox_to_anchor=(0.5, -0.08),
              fontsize=20, framealpha=0.9, ncol=2)

    plt.tight_layout()
    plt.savefig(path, dpi=300, bbox_inches="tight")
    plt.close(fig)

seen_path = "waymo_seen_vlc10_vs_fusionssd_radar_map.pdf"
unseen_path = "waymo_unseen_vlc10_vs_fusionssd_radar_map.pdf"

radar_plot(seen_path, seen_vlc10, "VLC Fusion", seen_fusionssd, "Fusion SSD", 50)
radar_plot(unseen_path, unseen_vlc10, "VLC Fusion", unseen_fusionssd, "Fusion SSD", 50)

print(seen_path)
print(unseen_path)
```
